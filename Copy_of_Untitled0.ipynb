```python
import urllib.request
from bs4 import BeautifulSoup

url = "https://www.wunderground.com/history/daily/us/hi/kailua-kona/date/2023-10-9"

# Request with a proper User-Agent header to avoid blocks
request = urllib.request.Request(url)
request.add_header("User-Agent", "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36")

# Connection and read the html content
opener = urllib.request.build_opener()
response = opener.open(request)
raw_html = response.read().decode("utf-8")

soup = BeautifulSoup(raw_html, 'html.parser')
```

```python
print(soup.prettify())
```

```python
ulnode = soup.find("ul", class_="list")
```

```python
# Find all tables in the webpage
tables = soup.find_all("table")

# Check that at least one table was found
if tables:

    # Select the first table
    table = tables[6]

    # Go through every row
    for row in table.find_all("tr"):

        # All heading and data cells
        cells = row.find_all(["td", "th"])


# Print every cell's text and divide up aesthetically
        print(" | ".join(cell.get_text(" ", strip=True) for cell in cells))
```

```python
import pandas as pd

# HTML tables in the site
tables = soup.find_all("table")

table = tables[6]

# list to store the rows
rows = []

# Scanning each row
for row in table.find_all("tr"):

# Headings and data cells in the row
 cells = row.find_all(["td", "th"])

# Store the text from each cell remove extras
if cells:
        rows.append([cell.get_text(" ", strip=True) for cell in cells])

# Use the first row as the DF column names
headers = rows[0]

# Use the remains rows as the table data
data = rows[1:]

# Create a DF from the scraped table
weather_df = pd.DataFrame(data, columns=headers)

display(weather_df)
```
