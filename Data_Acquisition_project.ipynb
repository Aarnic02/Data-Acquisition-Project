```python
# The brief said datasets had to be merged. Above is techicanlly not a merge even though all the weather data is the same for each athlete
#copy of Ironman DataFrame original stays same
#
ironman_weather_df = df.copy()

# The first weather record
weather_values = weather_df.iloc[0].to_dict()

# Adding all weather fields as a new column for each athelte athlete
for column, value in weather_values.items():
    ironman_weather_df[column] = value

# Display the merged
display(ironman_weather_df.head())

```

Issues faced while merging data:

- Different levels of detail: The weather dataset included hourly readings for the race day, whereas the Ironman dataset contained one row for each competitor. As a result, there was no direct one-to-one match between the meteorological data and each participant.

- Repeated weather values: The race-day weather data has to be appended to athlete rows in order to merge the datasets. This meant that several athletes saw the same weather values.

- Data cleaning was required because the athlete dataset contained time values that needed to be transformed into numerical minutes for analysis, and the scraped weather values included units like °F, %, mph, and in.

- Possible missing or inconsistent information: Scraped data may include missing or differently formatted values that should be verified before combining, and some athlete records had unknown divisions.

```python
# Import the libraries
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Create a copy to calculate values safely
plot_df = ironman_weather_df.copy()

# Convert overall_time (H:MM:SS) to total hours as float
def time_to_hours(time_str):
    if pd.isna(time_str) or not isinstance(time_str, str):
        return None
    try:
        parts = list(map(int, time_str.split(':')))
        if len(parts) == 3:
            return parts[0] + parts[1]/60.0 + parts[2]/3600.0
        return None
    except Exception:
        return None

plot_df['overall_time_hours'] = plot_df['overall_time'].apply(time_to_hours)

# Create a bar chart of average finish time by age group (div)
sns.barplot(data=plot_df, x="div", y="overall_time_hours")

# Add the scraped race-day temperature to the chart title
plt.title("Average Finish Time by Age Group | Race temperature: 25°C")
plt.ylabel("Overall Time (Hours)")
plt.xlabel("Division")

# Rotate the age-group labels so they are easier to read
plt.xticks(rotation=45)

# Display the chart
plt.show()
```

```python
# Race-Day Weather Parameter Trends Over Time

# Import the libraries used to work with data and create charts
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Make a copy of the scraped weather data so the original data is not changed
trends_df = weather_df.copy()

# Convert the time into a format Python can understand
# This makes it easier to sort the weather observations in time order
trends_df['Time_Parsed'] = pd.to_datetime(
    trends_df['Time'],
    format='%I:%M %p'
).dt.time

# Remove the units from each weather measurement and convert the values into numbers
# For example, '75 °F' becomes 75.0
trends_df['Temp_F'] = trends_df['Temperature'].str.replace(' °F', '', regex=False).astype(float)

# For example, '80 %' becomes 80.0
trends_df['Humidity_pct'] = trends_df['Humidity'].str.replace(' %', '', regex=False).astype(float)

# For example, '10 mph' becomes 10.0
trends_df['WindSpeed_mph'] = trends_df['Wind Speed'].str.replace(' mph', '', regex=False).astype(float)

# For example, '30 in' becomes 30.0
trends_df['Pressure_in'] = trends_df['Pressure'].str.replace(' in', '', regex=False).astype(float)

# Sort the rows so the earliest weather reading appears first
trends_df = trends_df.sort_values(by='Time_Parsed')

# Create a version of the time that is easier to read on the chart
# For example, the time will display as '08:00 AM'
trends_df['Time_Str'] = trends_df['Time_Parsed'].apply(
    lambda x: x.strftime('%I:%M %p')
)

# Create four charts arranged vertically
# Each chart will show a different weather measurement
# sharex=True means all four charts use the same time labels along the bottom
fig, axes = plt.subplots(4, 1, figsize=(12, 16), sharex=True)

# 1. Show how temperature changes throughout the day
sns.lineplot(
    data=trends_df,
    x='Time_Str',
    y='Temp_F',
    ax=axes[0],
    color='crimson',
    marker='o'
)
axes[0].set_title('Hourly Temperature Trend')  # Add a title to the chart
axes[0].set_ylabel('Temperature (°F)')        # Label the vertical axis
axes[0].grid(True, linestyle='--', alpha=0.5) # Add faint gridlines to help read values

# 2. Show how humidity changes throughout the day
sns.lineplot(
    data=trends_df,
    x='Time_Str',
    y='Humidity_pct',
    ax=axes[1],
    color='teal',
    marker='s'
)
axes[1].set_title('Hourly Humidity Trend')
axes[1].set_ylabel('Humidity (%)')
axes[1].grid(True, linestyle='--', alpha=0.5)

# 3. Show how wind speed changes throughout the day
sns.lineplot(
    data=trends_df,
    x='Time_Str',
    y='WindSpeed_mph',
    ax=axes[2],
    color='darkorange',
    marker='^'
)
axes[2].set_title('Hourly Wind Speed Trend')
axes[2].set_ylabel('Wind Speed (mph)')
axes[2].grid(True, linestyle='--', alpha=0.5)

# 4. Show how atmospheric pressure changes throughout the day
sns.lineplot(
    data=trends_df,
    x='Time_Str',
    y='Pressure_in',
    ax=axes[3],
    color='navy',
    marker='d'
)
axes[3].set_title('Atmospheric Pressure Trend')
axes[3].set_ylabel('Pressure (in)')
axes[3].grid(True, linestyle='--', alpha=0.5)

# Rotate the time labels so they do not overlap
plt.xticks(rotation=45, ha='right')

# Add a label underneath the charts to explain the time axis
plt.xlabel('Time of Day')

# Add one main title above all four charts
plt.suptitle(
    'Kailua Kona Race-Day Weather Conditions Overview',
    fontsize=16,
    weight='bold',
    y=1.01
)

# Adjust spacing so the titles and labels fit
plt.tight_layout()


plt.show()
```
